# Modules

The cells can import ES modules with the `import` statement, or with the `import()` function.

## Local modules

The paths of the modules are relative to the directory of the notebook.
`modules/geometry.mjs` exports some functions and a class:

```javascript
import { area, perimeter, Circle } from './modules/geometry.mjs';

[area({ width: 3, height: 4 }), perimeter({ width: 3, height: 4 }), new Circle(1).area()]
```

JSON files can be imported too, with an import attribute:

```javascript
import planets from './modules/planets.json' with { type: 'json' };

console.table(planets);
```

```javascript
const { Circle: C } = await import('./modules/geometry.mjs');
new C(2)
```

## The standard library of QuickJS

The [`std` and `os` modules](https://bellard.org/quickjs/quickjs.html#Standard-library) of QuickJS give
access to the files, the environment variables and the operating system.

```javascript
import * as std from 'std';
import * as os from 'os';

const file = std.open('hello.txt', 'w');
file.puts('Hello from QuickJS!\n');
file.printf('%d + %d = %d\n', 1, 2, 3);
file.close();

std.loadFile('hello.txt')
```

```javascript
const [entries] = os.readdir('.');
entries.filter(name => !name.startsWith('.')).sort()
```

```javascript
const [stat] = os.stat('hello.txt');
({ size: stat.size, modified: new Date(stat.mtime) })
```

```javascript
os.remove('hello.txt');
[os.platform, os.getcwd()[0].startsWith('/') ? 'absolute path' : 'relative path']
```

What QuickJS writes on its standard output, for instance with `std.printf`, is shown in the notebook:

```javascript
std.printf('%s has %d characters\n', 'QuickJS', 'QuickJS'.length);
std.out.puts('std.out.puts works too\n');
```
