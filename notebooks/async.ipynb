# Asynchronous JavaScript

The cells can use promises, `async` functions and timers. A cell runs until its code is done, its promises
are settled and its timers have fired, like a script with Node.js.

## Top-level await

```javascript
const sleep = ms => new Promise(resolve => setTimeout(resolve, ms));

const start = performance.now();
await sleep(500);
`Waited ${Math.round(performance.now() - start)} ms`
```

```javascript
async function fetchUser(id) {
  await sleep(200 * id);
  return { id, name: ['Ada', 'Grace', 'Alan'][id - 1] };
}

console.time('all users');
const users = await Promise.all([1, 2, 3].map(fetchUser));
console.timeEnd('all users');
users
```

## Timers

`setTimeout`, `setInterval`, `setImmediate` and `queueMicrotask` are available.

```javascript
setTimeout(() => console.log('timeout'), 0);
setImmediate(() => console.log('immediate'));
queueMicrotask(() => console.log('microtask'));
Promise.resolve().then(() => console.log('promise'));
console.log('synchronous');
```

```javascript
let ticks = 0;
const timer = setInterval(() => {
  ticks++;
  console.log(`tick ${ticks}`);
  if (ticks === 5) {
    clearInterval(timer);
  }
}, 200);
```

## Async generators

```javascript
async function* countdown(from) {
  for (let i = from; i > 0; i--) {
    await sleep(250);
    yield i;
  }
}

for await (const value of countdown(5)) {
  console.log(value);
}
```

## Errors

Errors in timers and unhandled promise rejections are reported in the cell which created them.

```javascript
setTimeout(() => {
  null.property;
}, 100);
```

```javascript
async function failing() {
  await sleep(100);
  throw new Error('Something failed later');
}

failing();
```

## Interrupting

Interrupt the kernel (with the stop button, or `I`, `I` in command mode) to stop an endless loop, or a timer
which never stops.

```javascript
let count = 0;
setInterval(() => {
  count++;
  clearOutput(true);
  console.log(`${count} ticks`);
}, 100);
```
