# xeus-quickjs

A Jupyter kernel for JavaScript, based on [QuickJS](https://bellard.org/quickjs/), the small and embeddable
JavaScript engine by Fabrice Bellard, and on [xeus](https://github.com/jupyter-xeus/xeus), the native
implementation of the Jupyter protocol.

QuickJS supports the latest ECMAScript features, like modules, async generators, BigInt and top-level await.

## Functions and classes

```javascript
/**
 * A point in the plane.
 */
class Point {
  constructor(x, y) {
    this.x = x;
    this.y = y;
  }

  /** The distance to the origin. */
  norm() {
    return Math.hypot(this.x, this.y);
  }
}

new Point(3, 4)
```

```javascript
new Point(3, 4).norm()
```

```javascript
const fibonacci = n => (n < 2 ? n : fibonacci(n - 1) + fibonacci(n - 2));
Array.from({ length: 15 }, (_, i) => fibonacci(i))
```

## Asynchronous code

`await` can be used at the top level of the cells, and the timers work like in a browser: a cell runs until
its promises are settled and its timers have fired.

```javascript
const sleep = ms => new Promise(resolve => setTimeout(resolve, ms));

console.log('Waiting...');
await sleep(1000);
'Done!'
```

```javascript
setTimeout(() => console.log('after 300ms'), 300);
setTimeout(() => console.log('after 100ms'), 100);
Promise.resolve().then(() => console.log('microtask'));
console.log('synchronous');
```

## Rich display

`display` shows HTML, Markdown, SVG, images, LaTeX or JSON. Objects can define their own rich representation
with methods like `_toHtml`. See the [rich display](rich-display.ipynb) notebook for more examples.

```javascript
display.markdown`**${engine}** supports *Markdown*, and LaTeX: $e^{i\pi} + 1 = 0$`;
display.html('<span style="color: tomato; font-size: 1.5em">HTML</span>');
```

```javascript
class Color {
  constructor(r, g, b) {
    Object.assign(this, { r, g, b });
  }

  _toHtml() {
    const { r, g, b } = this;
    return `<div style="width: 60px; height: 30px; border-radius: 4px; background: rgb(${r}, ${g}, ${b})"></div>`;
  }
}

new Color(70, 130, 180)
```

## Widgets

The Jupyter widgets are available in `Jupyter.widgets`. See the [widgets](widgets.ipynb) notebook.

```javascript
const { IntSlider, HTML, VBox } = Jupyter.widgets;

const slider = new IntSlider({ value: 4, min: 1, max: 12, description: 'Factorial' });
const result = new HTML();
const update = () => {
  let value = 1n;
  for (let i = 2n; i <= BigInt(slider.value); i++) {
    value *= i;
  }
  result.value = `<b>${slider.value}! = ${value}</b>`;
};
slider.observe(update, 'value');
update();

new VBox({ children: [slider, result] })
```
